```python
import numpy as np
```

```python
from ml_scratch.classifier.tree.DecisionTree import DecisionTreeClassifier
```

```python
X = np.array([
    [20,100],
    [22,110],
    [24,120],
    [45,400],
    [47,420],
    [50,450]
])

y = np.array([
    0,
    0,
    0,
    1,
    1,
    1
])
```

```python
def validate_pred(y_true : np.ndarray, y_pred : np.ndarray) -> float:
    
    correct_pred = y_true == y_pred
    accuracy = np.sum(correct_pred)/(correct_pred.size)
    print(f"Training Accuracy : {float(accuracy)*100:.1f}")
    return accuracy
```

```python
#Model 1 Depth 1
model = DecisionTreeClassifier(max_depth=1,min_sample_split=2,min_sample_leaf=1)
first_tree = model.fit(X,y)
model.print_tree(first_tree, ["Age", "Weight"])
pred=model.predict(X)
validate_pred(y, pred)
```

```python
#Model 2 Depth 2
model2 = DecisionTreeClassifier(max_depth=2,min_sample_split=2,min_sample_leaf=2)
second_tree = model2.fit(X,y)
model2.print_tree(second_tree, ["Age"])
pred=model2.predict(X)
validate_pred(y, pred)
```

```python
#Model 3 Depth 10
model3 = DecisionTreeClassifier(max_depth=10,min_sample_split=2,min_sample_leaf=2)
second_tree = model3.fit(X,y)
model3.print_tree(second_tree, ["Age"])
pred=model3.predict(X)
validate_pred(y, pred)
```

```python
#Model 4 Depth 10
model4 = DecisionTreeClassifier(max_depth=10,min_sample_split=2,min_sample_leaf=3)
second_tree = model4.fit(X,y)
model4.print_tree(second_tree, ["Age"])
pred=model4.predict(X)
validate_pred(y, pred)
```
